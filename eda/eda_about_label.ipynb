# 1. Import

```python
import pandas as pd
```

# 2. Configuration

```python
class Configuration:
    label_names = [
        'Left Infraclinoid Internal Carotid Artery',
        'Right Infraclinoid Internal Carotid Artery',
        'Left Supraclinoid Internal Carotid Artery',
        'Right Supraclinoid Internal Carotid Artery',
        'Left Middle Cerebral Artery',
        'Right Middle Cerebral Artery',
        'Anterior Communicating Artery',
        'Left Anterior Cerebral Artery',
        'Right Anterior Cerebral Artery',
        'Left Posterior Communicating Artery',
        'Right Posterior Communicating Artery',
        'Basilar Tip',
        'Other Posterior Circulation',
        'Aneurysm Present',
    ]
    
CFG = Configuration()
```

# 2. Load DataFrame

```python
train_df = pd.read_csv('../train.csv')
```

# 3. EDA

```python
total = []

for label_name in CFG.label_names:
    label_series = train_df[label_name]
    
    print(label_name)
    print('==================================================================')
    print('0 counts:', label_series.value_counts(dropna=False)[0])
    print('1 counts:', label_series.value_counts(dropna=False)[1])
    
    print('na counts:', label_series.isna().sum())
    
    total.append(sum(label_series == 1))
    
```

```python
total
```

```python
summary_df = pd.DataFrame({
    'total': total,
}, index=CFG.label_names)

summary_df
```

```python
weights = []

for label_name in CFG.label_names:
    positive = summary_df.loc[label_name, 'total']
    negative = len(train_df) - positive
    weights.append(negative / positive)
    
weights
```
