このNotebookは、画像データ(.npy)のファイルパスをDataFrameに収めて保存するためのものである。

# 1. Configuration

```python
class Configuration:
    
    image_size = 512
    
CFG = Configuration()
```

# 2. Import

```python
import os
from glob import glob
import pandas as pd
from tqdm.notebook import tqdm
```

```python
base_dir = f"../series_npy/{CFG.image_size}"
npy_path_df = pd.DataFrame()

# Iterate SeriesInstanceUIDs
for series_id in os.listdir(base_dir):

    series_path = base_dir + f'/{series_id}'
    npy_paths = glob(series_path + '/*.npy')
    npy_paths = [path.replace('\\', '/') for path in npy_paths]
    
    series_ids = [series_id] * len(npy_paths)
    
    temp_df = pd.DataFrame(
        {
            "series_id": series_ids,
            "npy_path": npy_paths
        },
        index=range(len(npy_paths))
    )
    npy_path_df = pd.concat([npy_path_df, temp_df], axis=0)
```

```python
os.makedirs('../npy_path/', exist_ok=True)
npy_path_df.to_csv(f'../npy_path/npy_{CFG.image_size}_path_df.csv', index=False)
```
